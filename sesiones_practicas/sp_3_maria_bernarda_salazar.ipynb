```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from scipy.stats import zscore

datos = load_wine()
df = pd.DataFrame(datos.data , columns=datos.feature_names)
df.head(8)
```

**Selección de variables númericas**
- alcohol
- color_intensity
- flavanoids

### **Aplicación de métodos de escalamiento**

**StandardScaler**

```python
#StandardScaler
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df[['alcohol', 'color_intensity', 'flavanoids']])
df_scaled = pd.DataFrame(df_scaled, columns=['alcohol', 'color_intensity', 'flavanoids'])
df_scaled.head(8)
```

**MinMaxScaler**

```python
# MinMaxScaler
minmax = MinMaxScaler()
df_minmax = minmax.fit_transform(df[['alcohol', 'color_intensity', 'flavanoids']])
df_minmax = pd.DataFrame(df_minmax, columns=['alcohol', 'color_intensity', 'flavanoids'])
df_minmax.head(8)
```

**RobustScaler**

```python
# RobustScaler
robust = RobustScaler()
df_robust = robust.fit_transform(df[['alcohol', 'color_intensity', 'flavanoids']])
df_robust = pd.DataFrame(df_robust, columns=['alcohol', 'color_intensity', 'flavanoids'])
df_robust.head(8)
```

**Z-Score**

```python
# Z-Score
df_zscore = zscore(df[['alcohol', 'color_intensity', 'flavanoids']])
df_zscore = pd.DataFrame(df_zscore, columns=['alcohol', 'color_intensity', 'flavanoids'])
df_zscore.head(8)
```

### Gráfica

```python
variable = 'flavanoids'

plt.figure(figsize=(15, 10))

# Grafico 1: Datos originales
plt.subplot(3, 2, 1)
sns.histplot(df[variable], kde=True)
plt.title(f'Distribución Original de {variable}')

# StandardScaler
plt.subplot(3, 2, 2)
sns.histplot(df_scaled[variable], kde=True)
plt.title(f'Distribución {variable} con StandardScaler')

# MinMaxScaler
plt.subplot(3, 2, 3)
sns.histplot(df_minmax[variable], kde=True)
plt.title(f'Distribución {variable} con MinMaxScaler')

# RobustScaler
plt.subplot(3, 2, 4)
sns.histplot(df_robust[variable], kde=True)
plt.title(f'Distribución {variable} con RobustScaler')

# Z-Score Distribution
plt.subplot(3, 2, 5)
sns.histplot(df_zscore[variable], kde=True)
plt.title(f'Distribución {variable} con Z-Score')

plt.tight_layout()
plt.show()
```

### Análisis

Según lo observado luego de realizar las gráficas de los diferentes metodos de escalamiento, la similitud entre StandardScaler y Z-Score demuestra que equivalencia debido a que la distribución de los datos no es muy distorsionada (no hay outliers fuertes). MinMaxScaler conserva la asimetría original, evidenciando que la variable no es naturalmente uniforme dentro de un rango acotado. En contraste, RobustScaler muestra una leve concentración de los datos centrales, indicando que existe cierto impacto de outliers que este método logra mitigar parcialmente.

### <span style="color:#2F749F;"><strong>Aplicación de Normalización L1 y L2</strong></span>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import Normalizer

datos_housing = fetch_california_housing()
df_housing = pd.DataFrame(datos_housing.data, columns=datos_housing.feature_names)

df_housing.head()
```

```python
vars_housing = ['MedInc', 'AveRooms', 'HouseAge']

df_housing_norm = df_housing[vars_housing].copy()

# 4. Aplicar Normalización L1
normalizer_l1 = Normalizer(norm='l1')
df_l1 = normalizer_l1.fit_transform(df_housing_norm)
df_l1 = pd.DataFrame(df_l1, columns=vars_housing)

print("Datos normalizados con L1:")
print(df_l1.head())

# Aplicar Normalización L2
normalizer_l2 = Normalizer(norm='l2')
df_l2 = normalizer_l2.fit_transform(df_housing_norm)
df_l2 = pd.DataFrame(df_l2, columns=vars_housing)

print("\nDatos normalizados con L2:")
print(df_l2.head())
```

### <span style="color:#2F749F;"><strong>Gráficas Comparativas</strong></span>

```python
# 5. Generar gráficas comparativas para una de las variables
variable_to_plot_housing = 'MedInc'

plt.figure(figsize=(15, 5))

# Distribución Original
plt.subplot(1, 3, 1)
sns.histplot(df_housing[variable_to_plot_housing], kde=True)
plt.title(f'Distribución Original de {variable_to_plot_housing}')

# Distribución con Normalización L1
plt.subplot(1, 3, 2)
sns.histplot(df_l1[variable_to_plot_housing], kde=True)
plt.title(f'Distribución {variable_to_plot_housing} con Normalización L1')

# Distribución con Normalización L2
plt.subplot(1, 3, 3)
sns.histplot(df_l2[variable_to_plot_housing], kde=True)
plt.title(f'Distribución {variable_to_plot_housing} con Normalización L2')

plt.tight_layout()
plt.show()
```

### <span style="color:#2F749F;"><strong>Análisis</strong></span>

La normalización L1 y L2 ajustan la magnitud de los vectores sin alterar la relación entre sus características, permitiendo comparabilidad entre observaciones. La normalización L1 tiende a generar representaciones más dispersas, lo que puede favorecer la interpretabilidad y la selección de características en modelos sensibles a la sparsidad. Por su parte, la normalización L2 distribuye el peso de manera más uniforme, preservando las proporciones relativas y siendo especialmente adecuada para algoritmos basados en distancia euclidiana. En términos de impacto, L2 suele ser más estable para modelos de aprendizaje automático como KNN o redes neuronales. En consecuencia, la elección entre ambas depende del tipo de modelo y del comportamiento esperado frente a la dispersión de los datos.

## <span style="color:#2F749F;"><strong>Ejercicio 4: Comparación de transformaciones para reducir asimetría</strong></span>

A partir de un conjunto de datos real sobre **pacientes con diabetes**, analiza cómo cambian las distribuciones de algunas variables clínicas al aplicar diferentes **transformaciones matemáticas utilizadas para reducir la asimetría (skewness) o estabilizar la varianza**.

1. Carga el conjunto de datos **diabetes** desde la librería `sklearn.datasets` usando `load_diabetes()`.

2. Convierte el conjunto de datos en un **DataFrame de pandas** e identifica las variables disponibles.

3. Selecciona al menos **dos variables numéricas** para el análisis (por ejemplo: `bmi`, `bp` o `s5`).

4. Aplica las siguientes **transformaciones** sobre una de las variables seleccionadas:

   - **Transformación logarítmica** usando `np.log()` o `np.log1p()`
   - **Transformación logarítmica reflejada** (útil cuando existe asimetría negativa)
   - **Transformación cuadrática** usando `x²`
   - **Transformación Box-Cox** usando `scipy.stats.boxcox`

5. Genera **gráficas comparativas** que permitan visualizar la distribución de la variable en los siguientes escenarios:
   - Distribución **original**
   - Distribución con **transformación log**
   - Distribución con **log reflejado**
   - Distribución con **transformación cuadrática**
   - Distribución con **transformación Box-Cox**

6. Escribe un análisis de **máximo cinco líneas** en el que expliques las diferencias entre las técnicas o métodos, su impacto en los resultados y cuál consideras más recomendable, justificando brevemente tu elección.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_diabetes
from scipy.stats import boxcox, skew

dataset = load_diabetes()
df_diabetes_ex4 = pd.DataFrame(dataset.data, columns=dataset.feature_names)


variable = 'bmi'


min_bmi = df_diabetes_ex4[variable].min()
if min_bmi <= 0:
    offset = np.abs(min_bmi) + 0.001
else:
    offset = 0

df_diabetes_ex4['bmi_adjusted'] = df_diabetes_ex4[variable] + offset

df_diabetes_ex4['bmi_log'] = np.log(df_diabetes_ex4['bmi_adjusted'])

# Transformación logarítmica reflejada (para asimetría negativa)
# Se refleja la variable restando el valor máximo + 1 y luego se aplica log
max_val = df_diabetes_ex4[variable].max()
df_diabetes_ex4['bmi_reflected_log'] = np.log(max_val + 1 - df_diabetes_ex4[variable])

# Transformación cuadrática
df_diabetes_ex4['bmi_quadratic'] = df_diabetes_ex4[variable]**2

# Transformación Box-Cox (requiere que los datos sean positivos)
df_diabetes_ex4['bmi_boxcox'], lambda_boxcox = boxcox(df_diabetes_ex4['bmi_adjusted'])

print("Datos después de transformaciones:")
display(df_diabetes_ex4[['bmi', 'bmi_log', 'bmi_reflected_log', 'bmi_quadratic', 'bmi_boxcox']].head())
```

### <span style="color:#2F749F;"><strong>Gráficas Comparativas</strong></span>

```python
plt.figure(figsize=(20, 12))

# Distribución original
plt.subplot(3, 2, 1)
sns.histplot(df_diabetes_ex4[variable], kde=True)
plt.title(f'Distribución Original de {variable} (Skew: {skew(df_diabetes_ex4[variable]):.2f})')

# Log Transformation
plt.subplot(3, 2, 2)
sns.histplot(df_diabetes_ex4['bmi_log'], kde=True)
plt.title(f'Distribución Log de {variable} (Skew: {skew(df_diabetes_ex4['bmi_log']):.2f})')

# Reflected Log Transformation
plt.subplot(3, 2, 3)
sns.histplot(df_diabetes_ex4['bmi_reflected_log'], kde=True)
plt.title(f'Distribución Log Reflejada de {variable} (Skew: {skew(df_diabetes_ex4['bmi_reflected_log']):.2f})')

# Quadratic Transformation
plt.subplot(3, 2, 4)
sns.histplot(df_diabetes_ex4['bmi_quadratic'], kde=True)
plt.title(f'Distribución Cuadrática de {variable} (Skew: {skew(df_diabetes_ex4['bmi_quadratic']):.2f})')

# Box-Cox Transformation
plt.subplot(3, 2, 5)
sns.histplot(df_diabetes_ex4['bmi_boxcox'], kde=True)
plt.title(f'Distribución Box-Cox de {variable} (Skew: {skew(df_diabetes_ex4['bmi_boxcox']):.2f})')

plt.tight_layout()
plt.show()
```

### <span style="color:#2F749F;"><strong>Análisis</strong></span>

Las transformaciones logarítmica y Box-Cox reducen la asimetría positiva al comprimir los valores extremos, mejorando la aproximación a la normalidad. La transformación logarítmica reflejada permite tratar distribuciones con asimetría negativa, ampliando su aplicabilidad. En contraste, la transformación cuadrática puede incrementar la asimetría o alterar la distribución de forma no deseada, por lo que su uso debe ser cuidadoso. Box-Cox destaca por su capacidad de optimizar automáticamente el parámetro de transformación, logrando un ajuste más preciso de la distribución. En términos de impacto, estas transformaciones facilitan el uso de modelos que asumen normalidad, siendo Box-Cox la opción más robusta cuando los datos son estrictamente positivos.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.preprocessing import PowerTransformer

heart_disease = fetch_ucirepo(id=45)

# Crear DataFrame
df_heart = pd.DataFrame(heart_disease.data.features)
df_heart.head()
```

```python
variables_numericas = ['chol', 'trestbps', 'thalach']
df_heart[variables_numericas].describe()

variable = 'chol'

df = df_heart[[variable]].dropna().copy()

# Raíz cuadrada
df['sqrt'] = np.sqrt(df[variable])

# Recíproca (evitar división por 0)
df['reciprocal'] = 1 / (df[variable] + 1e-8)

# Yeo-Johnson
pt = PowerTransformer(method='yeo-johnson')
df['yeo_johnson'] = pt.fit_transform(df[[variable]])

df.head()
```

### <span style="color:#2F749F;"><strong>Gráficas Comparativas</strong></span>

```python
plt.figure(figsize=(14,10))

# Original
plt.subplot(2,2,1)
sns.histplot(df[variable], kde=True)
plt.title('Distribución Original')

# Raíz cuadrada
plt.subplot(2,2,2)
sns.histplot(df['sqrt'], kde=True)
plt.title('Transformación Raíz Cuadrada')

# Recíproca
plt.subplot(2,2,3)
sns.histplot(df['reciprocal'], kde=True)
plt.title('Transformación Recíproca')

# Yeo-Johnson
plt.subplot(2,2,4)
sns.histplot(df['yeo_johnson'], kde=True)
plt.title('Transformación Yeo-Johnson')

plt.tight_layout()
plt.show()
```

### <span style="color:#2F749F;"><strong>Análisis</strong></span>

La transformación de raíz cuadrada atenúa moderadamente la asimetría sin modificar significativamente la estructura de la distribución. En contraste, la transformación recíproca introduce una compresión excesiva de los valores altos, lo que puede afectar la interpretabilidad y la relación entre variables. Por su parte, Yeo-Johnson logra una distribución más simétrica y cercana a la normalidad, adaptándose tanto a valores positivos como negativos. En términos de impacto, esta transformación mejora la adecuación de los datos para modelos estadísticos y de aprendizaje automático sensibles a la distribución. En consecuencia, Yeo-Johnson se considera la alternativa más robusta al ofrecer un equilibrio entre ajuste y estabilidad.
