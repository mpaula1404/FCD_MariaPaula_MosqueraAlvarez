## Correción de errores tipográficos


```python
import unicodedata

# Función para quitar tildes
def tildes(data):
    data = unicodedata.normalize('NFKD', data)
    return ''.join([c for c in data if not unicodedata.combining(c)])

# Normalizar texto
df['Ciudad_std'] = (
    df['Ciudad']
    .str.lower()                # pasar a minúsculas
    .apply(tildes)              # quitar tildes
    .str.strip()                # quitar espacios extra
)

# Corrección de errores
correcciones = {'bogata': 'bogota'}
df['Ciudad_std'] = df['Ciudad_std'].replace(correcciones)

# Mostrar resultados
df
```

# <span style="color:#2F749F;"><strong>Ejercicio 5. Imputación con KNN y regresión</strong></span>

## Identificación del DataFrame

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 30

df = pd.DataFrame({
  'Edad': np.random.randint(18, 60, n),
  'Promedio': np.round(np.random.uniform(2.0, 5.0, n), 2),
  'Horas_estudio': np.random.randint(0, 40, n),
  'Ingreso_familiar': np.random.randint(800000, 8000000, n),
  'Semestres_cursados': np.random.randint(1, 10, n)
 })

# Introducción de valores faltantes con distintos patrones
df.loc[[2, 5, 8, 15], 'Promedio'] = np.nan
df.loc[[1, 7, 10, 18], 'Horas_estudio'] = np.nan
df.loc[[3, 6, 14], 'Ingreso_familiar'] = np.nan
```

```python
df.info()
print('\n')
df.isna().sum() / len(df) * 100
```

```python
df
```

## Desarollo ejercicio

- ¿Qué variables tienen mayor proporción de datos faltantes?
- ¿Observas algún patrón estructurado o parecen aleatorios?

Según lo observado en el DataFrame es que las variables con mayor proporción de datos faltantes es Promedio y Horas_estudio. Verificando el dataset no se observa ningún patrón en relación con los datos faltantes de cada variable.

### Imputacion con KNN y Regresión

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.impute import KNNImputer

# Imputación con KNN
imputer = KNNImputer(n_neighbors=4)
data_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)

# -------- Máscara de valores imputados --------
mask_missing = df["Promedio"].isnull()

# figure
plt.figure(figsize=(8,6))
plt.scatter(df["Semestres_cursados"], df["Promedio"],
            color="gray", alpha=0.7, label="Datos disponibles")
plt.scatter(df.loc[mask_missing, "Semestres_cursados"],
            [0]*mask_missing.sum(),
            marker="x", color="black", s=100, label="Valores faltantes")

plt.xlabel("Semestres_cursados")
plt.ylabel("Promedio")
plt.title("Datos con valores faltantes")
plt.legend()
plt.grid(True)
plt.show()

# figure
plt.figure(figsize=(8,6))
plt.scatter(data_imputed.loc[~mask_missing, "Semestres_cursados"],
            data_imputed.loc[~mask_missing, "Promedio"],
            color="gray", alpha=0.7, label="Original")
plt.scatter(data_imputed.loc[mask_missing, "Semestres_cursados"],
            data_imputed.loc[mask_missing, "Promedio"],
            color="lightcoral", edgecolors="black", linewidths=1.2,
            s=100, label="Imputado (nuevo)")

plt.xlabel("Semestres_cursados")
plt.ylabel("Promedio")
plt.title("Valores imputados con KNN (K=3)")
plt.legend()
plt.grid(True)
plt.show()

# -------- Verificación --------
print("Valores faltantes antes de imputar:")
print(df.isnull().sum())

print("\nValores faltantes después de imputar:")
print(data_imputed.isnull().sum())
```

**Imputación por regresión**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


# Separar datos completos y faltantes
mask_missing = df["Promedio"].isnull()
X_train = df.loc[~mask_missing, ["Semestres_cursados"]]
y_train = df.loc[~mask_missing, "Promedio"]

X_missing = df.loc[mask_missing, ["Semestres_cursados"]]

# regresión lineal
reg = LinearRegression()
reg.fit(X_train, y_train)

# valores faltantes
data_imputed = df.copy()
data_imputed.loc[mask_missing, "Promedio"] = reg.predict(X_missing)

#plt.figure(figsize=(8,6))
plt.scatter(df["Semestres_cursados"], df["Promedio"],
            color="gray", alpha=0.7, label="Datos disponibles")
plt.scatter(df.loc[mask_missing, "Semestres_cursados"],
            [0]*mask_missing.sum(),
            marker="x", color="black", s=100, label="Valores faltantes")

plt.xlabel("Semestres_cursados")
plt.ylabel("Promedio")
plt.title("Datos con valores faltantes")
plt.legend()
plt.grid(True)
plt.show()

#
plt.figure(figsize=(8,6))
plt.scatter(data_imputed.loc[~mask_missing, "Semestres_cursados"],
            data_imputed.loc[~mask_missing, "Promedio"],
            color="gray", alpha=0.7, label="Original")
plt.scatter(data_imputed.loc[mask_missing, "Semestres_cursados"],
            data_imputed.loc[mask_missing, "Promedio"],
            color="lightcoral", edgecolors="black", linewidths=1.2,
            s=100, label="Imputado (regresión)")
x_line = np.linspace(df["Semestres_cursados"].min(), df["Semestres_cursados"].max(), 100).reshape(-1,1)
y_line = reg.predict(x_line)
plt.plot(x_line, y_line, color="blue", linestyle="--", label="Regresión lineal")

plt.xlabel("Semestres_cursados")
plt.ylabel("Promedio")
plt.title("Valores imputados con regresión lineal")
plt.legend()
plt.grid(True)
plt.show()

# -------- Verificación --------
print("Valores faltantes antes de imputar:")
print(df.isnull().sum())

print("\nValores faltantes después de imputar:")
print(data_imputed.isnull().sum())
```

**Análisis**

Según lo observado con las dos técnicas, en el KNN depende de la cantidad de vecinos cercanos que esten parametrizados en el algoritmo, debido a que esto cambia la posición en la que quedaría los datos luego de la imputación porque esta realizando la comparación de distancia con más puntos. Observando con K = 3 en KNN los resultados obtenidos en la imputación son muy parecidos al realizarlo con la tecnica de regresión.

# <span style="color:#2F749F;"><strong>Ejercicio 6. Imputación con LOCF y NOCB</strong></span>

## Identificación del Dataset

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 30

df = pd.DataFrame({
    'Fecha': pd.date_range(start='2023-01-01', periods=n, freq='D'),
    'Ventas': np.round(np.random.uniform(100, 500, n), 2),
    'Inventario': np.random.randint(10, 100, n),
    'Clientes': np.random.randint(1, 50, n)
})

# Introducción de valores faltantes
df.loc[[3, 4, 5, 12, 13], 'Ventas'] = np.nan
df.loc[[7, 8, 20], 'Inventario'] = np.nan
df.loc[[1, 2, 18, 19], 'Clientes'] = np.nan
df = df.sort_values('Fecha').reset_index(drop=True)
```

```python
df.info()
```

```python
df
```

## Análisis de Dataset

- **¿Qué variables presentan datos faltantes consecutivos?**
  - Las variables 'Ventas', 'Inventario' y 'Clientes' presentan datos faltantes consecutivos.

- **¿Por qué este tipo de patrón es relevante en series temporales?**
  - En series temporales, los datos faltantes consecutivos son muy relevantes porque representan períodos completos donde no se registraron datos. Esto puede indicar fallos en la recolección de datos, interrupciones del sistema o eventos significativos. Esto afecta directamente la capacidad de identificar tendencias, estacionalidad y patrones cíclicos, lo cual es relevante para el análisis.

- **¿Qué implicaciones tiene el orden de los datos para este tipo de imputación?**
  - El orden de los datos es fundamental para métodos de imputación basados en el tiempo, como LOCF (Last Observation Carried Forward) y NOCB (Next Observation Carried Backward). Si los datos no están ordenados cronológicamente, LOCF utilizará un valor de una observación que no es la correcta.

## Imputación con LOCF y NOCB

```python
import matplotlib.pyplot as plt

# Crear una copia del DataFrame para la imputación LOCF
df_locf = df.copy()

# Identificar los valores faltantes en 'Ventas' para resaltarlos en las gráficas
mask_missing_ventas = df['Ventas'].isnull()

# Gráfica de los datos originales de 'Ventas' con valores faltantes
plt.figure(figsize=(12, 6))
plt.plot(df['Fecha'], df['Ventas'], 'o-', color='gray', label='Datos disponibles')
plt.scatter(df.loc[mask_missing_ventas, 'Fecha'], [df['Ventas'].mean()] * mask_missing_ventas.sum(),
            color='red', marker='x', s=100, label='Valores faltantes (original)')
plt.title('Ventas Originales con Valores Faltantes')
plt.xlabel('Fecha')
plt.ylabel('Ventas')
plt.legend()
plt.grid(True)
plt.show()

# Aplicar imputación LOCF a la columna 'Ventas'
df_locf['Ventas'] = df_locf['Ventas'].fillna(method='ffill')

# Gráfica de los datos de 'Ventas' después de la imputación LOCF
plt.figure(figsize=(12, 6))
plt.plot(df['Fecha'], df['Ventas'], 'o-', color='gray', alpha=0.7, label='Datos Originales')
plt.plot(df_locf['Fecha'], df_locf['Ventas'], 'o-', color='darkgreen', label='Datos Imputados (LOCF)')
plt.scatter(df.loc[mask_missing_ventas, 'Fecha'], df_locf.loc[mask_missing_ventas, 'Ventas'],
            color='lightgreen', edgecolors='black', linewidths=1.5, s=100, label='Valores Imputados por LOCF')
plt.title('Ventas Imputadas con LOCF')
plt.xlabel('Fecha')
plt.ylabel('Ventas')
plt.legend()
plt.grid(True)
plt.show()

print("\nValores faltantes en el DataFrame después de imputación LOCF (solo 'Ventas' afectada para visualización):")
print(df_locf.isnull().sum())
```

**Imputacion con NOCB**

```python
# Crear una copia del DataFrame para la imputación NOCB
df_nocb = df.copy()

# Aplicar imputación NOCB a la columna 'Ventas'
df_nocb['Ventas'] = df_nocb['Ventas'].fillna(method='bfill')

# Gráfica de los datos de 'Ventas' después de la imputación NOCB
plt.figure(figsize=(12, 6))
plt.plot(df['Fecha'], df['Ventas'], 'o-', color='gray', alpha=0.7, label='Datos Originales')
plt.plot(df_nocb['Fecha'], df_nocb['Ventas'], 'o-', color='darkblue', label='Datos Imputados (NOCB)')
plt.scatter(df.loc[mask_missing_ventas, 'Fecha'], df_nocb.loc[mask_missing_ventas, 'Ventas'],
            color='skyblue', edgecolors='black', linewidths=1.5, s=100, label='Valores Imputados por NOCB')
plt.title('Ventas Imputadas con NOCB')
plt.xlabel('Fecha')
plt.ylabel('Ventas')
plt.legend()
plt.grid(True)
plt.show()

print("\nValores faltantes en el DataFrame después de imputación NOCB (solo 'Ventas' afectada para visualización):")
print(df_nocb.isnull().sum())
```

**¿Cuál método consideras más apropiado para este caso y por qué?**

Al observar las gráficas, tanto LOCF como NOCB rellenan los valores faltantes manteniendo la continuidad de la serie temporal. LOCF propaga el último valor conocido hacia adelante, mientras que NOCB propaga el siguiente valor conocido hacia atrás.

El más apropiado depende del contexto específico y de lo que se asuma sobre la naturaleza de los datos faltantes.

Considerando la naturaleza de las variables de este ejercicio (Ventas, Inventario, Clientes), donde los valores suelen estar influenciados por el comportamiento reciente, LOCF suele ser el método más comúnmente apropiado si no hay una razón específica para pensar que el futuro ya se conoce o influye en el pasado faltante. Es más conservador y representa mejor la continuidad del estado o proceso en ausencia de nuevas mediciones.
